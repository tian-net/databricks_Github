```sql
-- Consulta 1: visualizar los primeros 10 registros
SELECT
  `FECHA_CORTE;ITEM;FEDERACION;EVENTO;PAIS_EVENTO;CIUDAD_EVENTO;FECHA_INICIO;FECHA_FIN;DEPORTISTA;PUESTO;ESPECIALIDAD;COLECTIVO;DEPARTAMENTO;PROVINCIA;DISTRITO;UBIGEO;FECHA_PUBLICACION`
FROM
  `inteligencia_deportiva`.`deportes`.`deportistas_eventos`
LIMIT 10;
-- Consulta 2: cantidad total de registros
SELECT
  COUNT(*) as total_registros
FROM
  `inteligencia_deportiva`.`deportes`.`deportistas_eventos`;
-- Consulta 3: cantidad de registros por disciplina
SELECT
  SPLIT(
    `FECHA_CORTE;ITEM;FEDERACION;EVENTO;PAIS_EVENTO;CIUDAD_EVENTO;FECHA_INICIO;FECHA_FIN;DEPORTISTA;PUESTO;ESPECIALIDAD;COLECTIVO;DEPARTAMENTO;PROVINCIA;DISTRITO;UBIGEO;FECHA_PUBLICACION`,
    ';'
  )[2] AS disciplina_deportiva,
  COUNT(*) AS cantidad_registros
FROM
  `inteligencia_deportiva`.`deportes`.`deportistas_eventos`
GROUP BY
  disciplina_deportiva
ORDER BY
  cantidad_registros DESC
LIMIT 15;
--Consulta 4: resumen de valores nulos
WITH datos_parseados AS (
  SELECT
    SPLIT(
      `FECHA_CORTE;ITEM;FEDERACION;EVENTO;PAIS_EVENTO;CIUDAD_EVENTO;FECHA_INICIO;FECHA_FIN;DEPORTISTA;PUESTO;ESPECIALIDAD;COLECTIVO;DEPARTAMENTO;PROVINCIA;DISTRITO;UBIGEO;FECHA_PUBLICACION`,
      ';'
    ) AS campos
  FROM
    `inteligencia_deportiva`.`deportes`.`deportistas_eventos`
)
SELECT
  'COLECTIVO' AS campo,
  COUNT(
    CASE
      WHEN
        get(campos, 11) IS NULL
        OR TRIM(get(campos, 11)) = ''
      THEN
        1
    END
  ) AS valores_nulos_vacios,
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 11) IS NULL
            OR TRIM(get(campos, 11)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  ) AS porcentaje_nulos
FROM
  datos_parseados
UNION ALL
SELECT
  'PAIS_EVENTO',
  COUNT(
    CASE
      WHEN
        get(campos, 4) IS NULL
        OR TRIM(get(campos, 4)) = ''
      THEN
        1
    END
  ),
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 4) IS NULL
            OR TRIM(get(campos, 4)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  )
FROM
  datos_parseados
UNION ALL
SELECT
  'CIUDAD_EVENTO',
  COUNT(
    CASE
      WHEN
        get(campos, 5) IS NULL
        OR TRIM(get(campos, 5)) = ''
      THEN
        1
    END
  ),
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 5) IS NULL
            OR TRIM(get(campos, 5)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  )
FROM
  datos_parseados
UNION ALL
SELECT
  'FECHA_INICIO',
  COUNT(
    CASE
      WHEN
        get(campos, 6) IS NULL
        OR TRIM(get(campos, 6)) = ''
      THEN
        1
    END
  ),
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 6) IS NULL
            OR TRIM(get(campos, 6)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  )
FROM
  datos_parseados
UNION ALL
SELECT
  'FECHA_FIN',
  COUNT(
    CASE
      WHEN
        get(campos, 7) IS NULL
        OR TRIM(get(campos, 7)) = ''
      THEN
        1
    END
  ),
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 7) IS NULL
            OR TRIM(get(campos, 7)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  )
FROM
  datos_parseados
UNION ALL
SELECT
  'DEPORTISTA',
  COUNT(
    CASE
      WHEN
        get(campos, 8) IS NULL
        OR TRIM(get(campos, 8)) = ''
      THEN
        1
    END
  ),
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 8) IS NULL
            OR TRIM(get(campos, 8)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  )
FROM
  datos_parseados
UNION ALL
SELECT
  'PUESTO',
  COUNT(
    CASE
      WHEN
        get(campos, 9) IS NULL
        OR TRIM(get(campos, 9)) = ''
      THEN
        1
    END
  ),
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 9) IS NULL
            OR TRIM(get(campos, 9)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  )
FROM
  datos_parseados
UNION ALL
SELECT
  'ESPECIALIDAD',
  COUNT(
    CASE
      WHEN
        get(campos, 10) IS NULL
        OR TRIM(get(campos, 10)) = ''
      THEN
        1
    END
  ),
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 10) IS NULL
            OR TRIM(get(campos, 10)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  )
FROM
  datos_parseados
UNION ALL
SELECT
  'DEPARTAMENTO',
  COUNT(
    CASE
      WHEN
        get(campos, 12) IS NULL
        OR TRIM(get(campos, 12)) = ''
      THEN
        1
    END
  ),
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 12) IS NULL
            OR TRIM(get(campos, 12)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  )
FROM
  datos_parseados
UNION ALL
SELECT
  'PROVINCIA',
  COUNT(
    CASE
      WHEN
        get(campos, 13) IS NULL
        OR TRIM(get(campos, 13)) = ''
      THEN
        1
    END
  ),
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 13) IS NULL
            OR TRIM(get(campos, 13)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  )
FROM
  datos_parseados
UNION ALL
SELECT
  'DISTRITO',
  COUNT(
    CASE
      WHEN
        get(campos, 14) IS NULL
        OR TRIM(get(campos, 14)) = ''
      THEN
        1
    END
  ),
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 14) IS NULL
            OR TRIM(get(campos, 14)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  )
FROM
  datos_parseados
UNION ALL
SELECT
  'UBIGEO',
  COUNT(
    CASE
      WHEN
        get(campos, 15) IS NULL
        OR TRIM(get(campos, 15)) = ''
      THEN
        1
    END
  ),
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 15) IS NULL
            OR TRIM(get(campos, 15)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  )
FROM
  datos_parseados
UNION ALL
SELECT
  'FECHA_PUBLICACION',
  COUNT(
    CASE
      WHEN
        get(campos, 16) IS NULL
        OR TRIM(get(campos, 16)) = ''
      THEN
        1
    END
  ),
  ROUND(
    100.0
      * COUNT(
        CASE
          WHEN
            get(campos, 16) IS NULL
            OR TRIM(get(campos, 16)) = ''
          THEN
            1
        END
      )
      / COUNT(*),
    2
  )
FROM
  datos_parseados
ORDER BY
  porcentaje_nulos DESC;
-- Consulta 5: eventos con mayor número de registros
SELECT
  get(
    SPLIT(
      `FECHA_CORTE;ITEM;FEDERACION;EVENTO;PAIS_EVENTO;CIUDAD_EVENTO;FECHA_INICIO;FECHA_FIN;DEPORTISTA;PUESTO;ESPECIALIDAD;COLECTIVO;DEPARTAMENTO;PROVINCIA;DISTRITO;UBIGEO;FECHA_PUBLICACION`,
      ';'
    ),
    3
  ) AS evento,
  get(
    SPLIT(
      `FECHA_CORTE;ITEM;FEDERACION;EVENTO;PAIS_EVENTO;CIUDAD_EVENTO;FECHA_INICIO;FECHA_FIN;DEPORTISTA;PUESTO;ESPECIALIDAD;COLECTIVO;DEPARTAMENTO;PROVINCIA;DISTRITO;UBIGEO;FECHA_PUBLICACION`,
      ';'
    ),
    2
  ) AS federacion,
  COUNT(*) AS cantidad_registros
FROM
  `inteligencia_deportiva`.`deportes`.`deportistas_eventos`
GROUP BY
  evento,
  federacion
ORDER BY
  cantidad_registros DESC
LIMIT 15
```
