```sql

-- 1. Crear el catálogo
CREATE CATALOG IF NOT EXISTS inteligencia_deportiva;

-- 2. Crear el schema
CREATE SCHEMA IF NOT EXISTS inteligencia_deportiva.deportes;

-- 3. Seleccionar el catálogo y schema
USE CATALOG inteligencia_deportiva;
USE SCHEMA deportes;

-- 4. Crear una tabla externa que lea el CSV de S3
CREATE TABLE IF NOT EXISTS deportistas_eventos
USING CSV
OPTIONS (
  header = 'true',
  inferSchema = 'true'
)
LOCATION 's3://fia-practica/Data_DeportistasEventos_0.csv';

-- 5. Consultar los primeros registros
SELECT *
FROM inteligencia_deportiva.deportes.deportistas_eventos
LIMIT 10;



DESCRIBE TABLE inteligencia_deportiva.deportes.deportistas_eventos;

SELECT COUNT(*) AS total_registros
FROM inteligencia_deportiva.deportes.deportistas_eventos;

```
